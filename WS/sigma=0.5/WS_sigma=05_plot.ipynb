```python
import networkx as nx
import random
import matplotlib.pyplot as plt
import numpy as np
import json
import warnings
warnings.filterwarnings('ignore')
```

```python
font = {'family': 'Times New Roman',
        'weight': 'bold',
        'style':'italic',
        'size': 26,
        }

font2 = {'family': 'Times New Roman',
        'weight': 'normal',
        'size': 22,
        }
colors=["#0000FF","#000000","#FF0033","#00CC00","#ff7500",'#56004f']
ETime = 10000
  
rlist = [0.1, 0.3, 0.5, 0.7, 1.0]
#f = open('./saves/t_f_RE.json')
f = open(r'C:\Users\22159\Desktop\论文\WS\sigma=0.5\t_f_WS_05.json')
f_c = json.load(f)
marker_list = ['o','^','s','D','H','>','<']
Y = []
X = list(range(ETime+2))
for i in range(5):
    Y.append(f_c['c={:.1f},r={:.1f}'.format(-rlist[i],1+rlist[i])])
plt.figure(figsize = (11, 10))
for i in range(5):
    plt.plot(X, Y[i],marker=marker_list[i],label=chr(114)+'='+str(1+rlist[i]),color=colors[i],lw=2.5,markersize=5,markerfacecolor="w")
plt.xlabel('$t$',fontdict=font)
plt.ylabel('$f_{c}$',fontdict=font)
plt.xticks(fontsize=20)
plt.yticks(fontsize=20)
plt.xscale('log')
plt.ylim([0,1])
plt.xlim([1,10000])
plt.legend(frameon=False, loc='best', handlelength=2, prop=font2)
#plt.savefig('t_f(RE).pdf', dpi = 600)
#plt.savefig('t_f(WS).pdf', dpi = 600)
plt.savefig('WS 05.jpg', dpi=300)
plt.show()
```
